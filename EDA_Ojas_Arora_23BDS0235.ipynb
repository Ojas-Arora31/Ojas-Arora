# Checking Missing Values

Missing values are identified using the isnull() function.

Handling missing values is an important step in data cleaning.

```python
df.isnull().sum()
```

# Data Cleaning

Data cleaning is performed to remove errors and handle missing values.

Numerical columns are filled using mean values.

Categorical columns are filled using mode values.

```python

df[df.select_dtypes(include=np.number).columns] = (
    df.select_dtypes(include=np.number)
    .fillna(df.select_dtypes(include=np.number).mean())
)

df[df.select_dtypes(include="object").columns] = (
    df.select_dtypes(include="object")
    .fillna(df.select_dtypes(include="object").mode().iloc[0])
)

df.isnull().sum()
```

# Verify Missing Values After Cleaning

Checking again to confirm that missing values are handled successfully.

```python
df.isnull().sum()
```

# Data Transformation

Machine learning algorithms require numerical data.

Categorical variables are converted into numerical values using Label Encoding.

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df_encoded = df.copy()

for col in df_encoded.select_dtypes(include="object").columns:
    df_encoded[col] = encoder.fit_transform(df_encoded[col])

df_encoded.head()
```

# Univariate Analysis

Univariate analysis is the analysis of a single variable.

It is used to understand:

- Distribution of data
- Central tendency
- Spread of values

Different visualization techniques are used to analyze individual variables.

```python
df_encoded.hist(figsize=(12,10))

plt.show()
```

# Univariate Analysis using Box Plot

Box plot is used to understand data distribution and detect outliers.

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df_encoded)

plt.xticks(rotation=90)

plt.show()
```

# Univariate Analysis using Distribution Plot

Distribution plot shows how values are distributed in a variable.

```python
plt.figure(figsize=(8,5))

sns.histplot(df_encoded.iloc[:,0], kde=True)

plt.show()
```

# Bivariate Analysis

Bivariate analysis studies the relationship between two variables.

It helps to understand correlation and dependency between variables.

```python
plt.figure(figsize=(7,5))

sns.scatterplot(
    x=df_encoded.iloc[:,0],
    y=df_encoded.iloc[:,1]
)

plt.show()
```

# Bivariate Analysis using Regression Plot

Regression plot shows the relationship and trend between two numerical variables.

```python
plt.figure(figsize=(7,5))

sns.regplot(
    x=df_encoded.iloc[:,0],
    y=df_encoded.iloc[:,1]
)

plt.show()
```

# Correlation Between Two Variables

Correlation coefficient represents the strength of relationship between two variables.

The value ranges between -1 and +1.

```python
np.corrcoef(
    df_encoded.iloc[:,0],
    df_encoded.iloc[:,1]
)
```

# Multivariate Analysis

Multivariate analysis involves analyzing three or more variables.

Pair plots are used to visualize relationships among multiple variables.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(df.select_dtypes(include='number'))
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(df.select_dtypes(include='number').corr(),
            annot=True,
            cmap='coolwarm')

plt.show()
```

# Correlation Matrix

Correlation matrix shows relationships among all numerical variables.

Pearson correlation method is used.

```python
correlation = df_encoded.corr()

correlation
```

# Heatmap Visualization

Heatmap provides a visual representation of correlation values.

Values close to +1 indicate strong positive correlation.

Values close to -1 indicate strong negative correlation.

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.show()
```
