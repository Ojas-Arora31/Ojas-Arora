### Arithmetic Mean

The Arithmetic Mean represents the average value of the observations for each numerical clinical variable.

It is calculated using:

\[
\bar{x} = \frac{\sum x_i}{n}
\]

where \(x_i\) represents each observation and \(n\) represents the total number of observations.

```python
# Arithmetic Mean

arithmetic_mean = clinical.mean()

print("Arithmetic Mean:")
display(arithmetic_mean)
```

Among the selected variables, pulse and total protein show relatively larger numerical values, while rectal temperature and abdomo protein have smaller numerical values. These mean values provide a numerical summary of the central tendency of the clinical variables.

### Weighted Arithmetic Mean

The **Weighted Arithmetic Mean** calculates the average by assigning different weights to observations.

It is given by:

\[
\bar{x}_w = \frac{\sum w_i x_i}{\sum w_i}
\]

where \(x_i\) represents an observation and \(w_i\) represents its corresponding weight.

For this analysis, the weighted mean will be calculated using observation-based weights.

### Quadratic Mean

The **Quadratic Mean (QM)**, also known as the **Root Mean Square (RMS)**, is calculated by taking the square root of the mean of the squared observations.

It is given by:

\[
QM = \sqrt{\frac{\sum_{i=1}^{n}x_i^2}{n}}
\]

The Quadratic Mean gives greater importance to larger values because each observation is squared before calculating the mean.

```python
# Quadratic Mean

quadratic_mean = np.sqrt((clinical ** 2).mean())

print("Quadratic Mean:")
display(quadratic_mean)
```

### Observation

The Quadratic Mean values are slightly higher than the corresponding arithmetic mean values for most variables, because the quadratic mean gives greater influence to larger observations.

### Harmonic Mean

The **Harmonic Mean** is a type of average that gives greater importance to smaller values.

It is calculated as:

\[
HM = \frac{n}{\sum_{i=1}^{n}\frac{1}{x_i}}
\]

The Harmonic Mean is calculated for positive observations of each numerical clinical variable.

```python
# Harmonic Mean

harmonic_mean = clinical.apply(
    lambda x: stats.hmean(x[x > 0])
)

print("Harmonic Mean:")
display(harmonic_mean)
```

### Observation

### Geometric Mean

The **Geometric Mean (GM)** is calculated as:

**GM = (x₁ × x₂ × x₃ × ... × xₙ)^(1/n)**

where:
- **x₁, x₂, ..., xₙ** are the observations.
- **n** is the total number of observations.

```python
# Geometric Mean

geometric_mean = clinical.apply(
    lambda x: stats.gmean(x[x > 0])
)

print("Geometric Mean:")
display(geometric_mean)
```

### Observation

The Geometric Mean values are generally lower than the corresponding Arithmetic Mean values because the Geometric Mean is less influenced by extremely large observations.

- **Rectal temperature:** 38.163003

The difference is particularly noticeable for **total protein**, where the Geometric Mean is considerably lower than the Arithmetic Mean. This indicates that larger observations have a stronger effect on the Arithmetic Mean than on the Geometric Mean.

### Weighted Geometric Mean

The **Weighted Geometric Mean** assigns different weights to observations when calculating the geometric mean.

It is given by:

**Weighted GM = (x₁^w₁ × x₂^w₂ × x₃^w₃ × ... × xₙ^wₙ)^(1 / (w₁ + w₂ + ... + wₙ))**

where:
- **x₁, x₂, ..., xₙ** are the observations.
- **w₁, w₂, ..., wₙ** are their corresponding weights.
- The sum of the weights is used as the denominator of the exponent.

## 2.1.3 Measures of Dispersion

Measures of dispersion describe the amount of variability or spread present in the numerical clinical data.

- Range

- Quartile Deviation
- Interdecile Range

These measures provide information about how widely the clinical observations are distributed around their central values.

### Range

The **Range** measures the total spread of a variable by calculating the difference between its maximum and minimum values.

**Range = Maximum Value − Minimum Value**

A larger range indicates greater variability in the observations.

```python
# 2.1.3 MEASURES OF DISPERSION
# Range


data_range = clinical.max() - clinical.min()

print("Range:")
display(data_range)
```

### Observation

### Interquartile Range (IQR)

The **Interquartile Range (IQR)** measures the spread of the middle 50% of observations.

It is calculated as:

**IQR = Q3 − Q1**

where:
- **Q1** is the 25th percentile.
- **Q3** is the 75th percentile.

The IQR is less affected by extreme values than the range and therefore provides a robust measure of dispersion.

```python
# Interquartile Range (IQR)

Q1 = clinical.quantile(0.25)
Q3 = clinical.quantile(0.75)

iqr = Q3 - Q1

print("Interquartile Range (IQR):")
display(iqr)
```

The **total protein** variable has the largest IQR, indicating substantial variability within its middle 50% of observations. **Pulse** also shows considerable dispersion. In contrast, **abdomo protein** has an IQR of 0.0 because its first and third quartiles are both 3.039604 in the cleaned dataset.

### Quartile Deviation

The **Quartile Deviation (QD)**, also known as the semi-interquartile range, measures the spread around the median by taking half of the difference between the third and first quartiles.

**Quartile Deviation = (Q3 − Q1) / 2**

It represents half of the Interquartile Range and is less affected by extreme observations.

```python
# Quartile Deviation


quartile_deviation = (Q3 - Q1) / 2

print("Quartile Deviation:")
display(quartile_deviation)
```

### Observation

The Quartile Deviation represents half of the Interquartile Range and shows the spread of the middle 50% of observations around the median.

### Interdecile Range

The **Interdecile Range (IDR)** measures the spread of the middle 80% of observations.

It is calculated as:

**Interdecile Range = D9 − D1**

where:

- **D1** is the 10th percentile.
- **D9** is the 90th percentile.

It provides a measure of dispersion that excludes the lowest 10% and highest 10% of observations.

```python
# Interdecile Range


D1 = clinical.quantile(0.10)
D9 = clinical.quantile(0.90)

interdecile_range = D9 - D1

print("Interdecile Range:")
display(interdecile_range)
```

### Mean Deviation from Mean

The **Mean Deviation from Mean** measures the average absolute distance of each observation from the arithmetic mean.

It is calculated as:

**Mean Deviation = Σ|xᵢ − x̄| / n**

where:

- **xᵢ** = individual observation
- **x̄** = arithmetic mean
- **n** = total number of observations

A higher value indicates greater dispersion around the mean.

```python
# Mean Deviation from Mean


mean_deviation_mean = clinical.apply(
    lambda x: np.mean(np.abs(x - x.mean()))
)

print("Mean Deviation from Mean:")
display(mean_deviation_mean)
```

A **Frequency Polygon** is a graphical representation of a frequency distribution. It is created by plotting the frequency of each class interval against its class midpoint and connecting the points with straight lines.

### Smoothed Frequency Curve

A **Smoothed Frequency Curve** is a smooth graphical representation of a frequency distribution. It is obtained by smoothing the points of the frequency polygon to show the overall shape and trend of the distribution.

It helps identify the concentration, spread, and general pattern of the numerical clinical variables.

```python
# Smoothed Frequency Curve


import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import make_interp_spline

for col in clinical.columns:
    counts, bins = np.histogram(clinical[col], bins=10)

    # Calculate class midpoints
    midpoints = (bins[:-1] + bins[1:]) / 2

    # Create smooth x-values
    x_smooth = np.linspace(midpoints.min(), midpoints.max(), 300)

    # Create smooth curve
    spline = make_interp_spline(midpoints, counts, k=2)
    y_smooth = spline(x_smooth)

    # Prevent negative frequencies
    y_smooth = np.maximum(y_smooth, 0)

    plt.figure(figsize=(8, 5))
    plt.plot(x_smooth, y_smooth)

    plt.title(f"Smoothed Frequency Curve - {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.grid(True, alpha=0.3)
    plt.show()
```

### Ogive

An **Ogive** is a cumulative frequency curve used to represent the cumulative distribution of numerical data.

It is constructed by plotting **class boundaries** against **cumulative frequencies**. An ogive helps identify how observations accumulate across different values and can also be used to estimate median, quartiles, and other positional measures.

```python
# Ogive - Less Than Cumulative Frequency


import numpy as np
import matplotlib.pyplot as plt

for col in clinical.columns:


    counts, bins = np.histogram(clinical[col], bins=10)


    cumulative_freq = np.cumsum(counts)

    upper_boundaries = bins[1:]

    plt.figure(figsize=(8, 5))

    plt.plot(
        upper_boundaries,
        cumulative_freq,
        marker='o'
    )

    plt.title(f"Ogive - {col}")
    plt.xlabel(f"{col} (Upper Class Boundary)")
    plt.ylabel("Cumulative Frequency")
    plt.grid(True, alpha=0.3)

    plt.show()
```

### Smoothed Ogive

A **Smoothed Ogive** is a smooth version of a cumulative frequency curve. It represents the cumulative frequency distribution while reducing sharp changes between consecutive class boundaries.

It helps visualize the overall cumulative pattern of the numerical clinical variables and shows how observations accumulate as the variable values increase.

```python
# Smoothed Ogive


import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import make_interp_spline

for col in clinical.columns:


    counts, bins = np.histogram(clinical[col], bins=10)


    cumulative_freq = np.cumsum(counts)


    upper_boundaries = bins[1:]


    x_smooth = np.linspace(
        upper_boundaries.min(),
        upper_boundaries.max(),
        300
    )


    spline = make_interp_spline(
        upper_boundaries,
        cumulative_freq,
        k=2
    )

    y_smooth = spline(x_smooth)

    y_smooth = np.clip(
        y_smooth,
        0,
        len(clinical)
    )

    plt.figure(figsize=(8, 5))

    plt.plot(
        x_smooth,
        y_smooth
    )

    plt.title(f"Smoothed Ogive - {col}")
    plt.xlabel(f"{col} (Upper Class Boundary)")
    plt.ylabel("Cumulative Frequency")
    plt.grid(True, alpha=0.3)

    plt.show()
```

### Expected Frequencies

**Expected Frequency** is the frequency that would be expected for each cell of a contingency table if the two categorical variables were statistically independent.

It is calculated as:

**Expected Frequency = (Row Total × Column Total) / Grand Total**

Expected frequencies are required for calculating the **Pearson Chi-Square statistic** and testing the independence between surgery and outcome.

```python
# Expected Frequencies


from scipy.stats import chi2_contingency

chi2, p_value, degrees_of_freedom, expected = chi2_contingency(
    contingency_table
)

expected_frequencies = pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
)

display(expected_frequencies.round(2))
```

  - Lived: **107.16**

Comparing the observed and expected frequencies shows noticeable differences in several cells. For example, the observed frequency for **surgery + died** is 58, compared with an expected frequency of 46.35.

These differences will be quantified using **Chi-Square Contributions** in the next step.

### Chi-Square Contributions

The **Chi-Square Contribution** measures how much each cell of a contingency table contributes to the overall Pearson Chi-Square statistic.

It is calculated as:

**Chi-Square Contribution = (Observed Frequency − Expected Frequency)² / Expected Frequency**

A larger contribution indicates a greater difference between the observed and expected frequencies for that cell.

```python
# Chi-Square Contributions


observed = contingency_table

chi_square_contributions = (
    (observed - expected_frequencies) ** 2
    / expected_frequencies
)

display(chi_square_contributions.round(4))
```

### Observation

The Chi-Square Contributions show how much each cell contributes to the overall Pearson Chi-Square statistic.

- **No surgery + euthanized:** Contribution = **0.0150**, a very small contribution.
- **No surgery + lived:** Contribution = **2.0863**.
- **Surgery + died:** Contribution = **2.9257**.
- **Surgery + euthanized:** Contribution = **0.0099**, the smallest contribution.
- **Surgery + lived:** Contribution = **1.3793**.

The largest contribution comes from the **no surgery + died** cell, indicating the greatest difference between its observed and expected frequencies. The euthanized categories contribute very little to the overall chi-square statistic.

### Pearson Chi-Square

The **Pearson Chi-Square statistic** measures the overall difference between the observed and expected frequencies in a contingency table.

It is calculated by summing all individual Chi-Square Contributions:

**Pearson Chi-Square = Σ [(Observed − Expected)² / Expected]**

A larger Chi-Square value indicates a greater overall difference between the observed and expected frequencies.

```python
# Pearson Chi-Square


pearson_chi_square = chi_square_contributions.to_numpy().sum()

print("Pearson Chi-Square:", round(pearson_chi_square, 4))
```

### Degrees of Freedom

Therefore:

The degrees of freedom are required for interpreting the Pearson Chi-Square test.

### Independence Test

The **Chi-Square Test of Independence** determines whether two categorical variables are statistically independent.

For this analysis, the variables are:

- **Surgery**
- **Outcome**

The hypotheses are:

**H₀:** Surgery and outcome are independent.

**H₁:** Surgery and outcome are not independent.

The test uses the **Pearson Chi-Square statistic** and its corresponding **p-value**.

A significance level of **α = 0.05** is commonly used:

- **p-value < 0.05:** Reject H₀.
- **p-value ≥ 0.05:** Do not reject H₀.

```python
# Independence Test


from scipy.stats import chi2_contingency

chi2_stat, p_value, df_test, expected_test = chi2_contingency(
    contingency_table
)

print("Pearson Chi-Square:", round(chi2_stat, 4))
print("Degrees of Freedom:", df_test)
print("p-value:", round(p_value, 6))

alpha = 0.05

if p_value < alpha:
    print("Decision: Reject H₀")
    print("Conclusion: Surgery and outcome are not independent.")
else:
    print("Decision: Do not reject H₀")
    print("Conclusion: There is insufficient evidence to conclude that surgery and outcome are associated.")
```

### Phi Coefficient

The **Phi (φ) coefficient** measures the strength of association between two categorical variables in a **2 × 2 contingency table**.

The current **surgery × outcome** table is **2 × 3**, so the Phi coefficient is **not directly appropriate** for the full table.

Therefore, for the 2 × 3 contingency table, the appropriate association measures are **Cramer's V** and **Tschuprow's T**.

To keep the analysis statistically correct, we will calculate **Cramer's V** next rather than forcing a Phi coefficient onto the 2 × 3 table.

```python
# Phi Coefficient


rows, columns = contingency_table.shape

if rows == 2 and columns == 2:
    phi = np.sqrt(chi2_stat / contingency_table.values.sum())
    print("Phi Coefficient:", round(phi, 4))
else:
    print("Phi Coefficient is not directly applicable.")
    print(f"Current contingency table size: {rows} x {columns}")
    print("Use Cramer's V for this 2 x 3 table.")
```

### Cramer's V

**Cramer's V** measures the strength of association between two categorical variables and is suitable for contingency tables larger than 2 × 2.

It is calculated as:

**V = √(χ² / [n × min(r − 1, c − 1)])**

where:

- **χ²** = Pearson Chi-Square statistic
- **n** = total number of observations
- **r** = number of rows
- **c** = number of columns

For the current **2 × 3 surgery × outcome** table, Cramer's V is appropriate for measuring the strength of association.

```python
# Cramer's V


n = contingency_table.to_numpy().sum()
r, c = contingency_table.shape

cramers_v = np.sqrt(
    chi2_stat / (n * min(r - 1, c - 1))
)

print("Cramer's V:", round(cramers_v, 4))
```

### Tschuprow's T

**Tschuprow's T** measures the strength of association between two categorical variables in a contingency table.

It is calculated as:

**T = √(χ² / [n × √((r − 1)(c − 1))])**

where:

- **χ²** = Pearson Chi-Square statistic
- **n** = total number of observations
- **r** = number of rows
- **c** = number of columns

For the current **2 × 3 surgery × outcome** contingency table, Tschuprow's T provides another measure of the strength of association.

```python
# Tschuprow's T


n = contingency_table.to_numpy().sum()
r, c = contingency_table.shape

tschuprow_t = np.sqrt(
    chi2_stat / (
        n * np.sqrt((r - 1) * (c - 1))
    )
)

print("Tschuprow's T:", round(tschuprow_t, 4))
```

**Regression Analysis** is used to model the relationship between a dependent numerical variable and an independent numerical variable.

For the Horse Dataset:

- **Independent variable (X):** Pulse
- **Dependent variable (Y):** Respiratory Rate

A simple linear regression model can be represented as:

**Y = β₀ + β₁X**

where:
- **Y** is the predicted respiratory rate.
- **X** is the pulse.
- **β₀** is the intercept.
- **β₁** is the regression coefficient.

The regression analysis helps quantify the linear relationship between pulse and respiratory rate.

```python
# Regression Analysis


from scipy.stats import linregress

x = clinical['pulse']
y = clinical['respiratory_rate']

regression_result = linregress(x, y)

slope = regression_result.slope
intercept = regression_result.intercept
r_value = regression_result.rvalue
p_value = regression_result.pvalue
std_err = regression_result.stderr

print("Slope:", round(slope, 4))
print("Intercept:", round(intercept, 4))
print("R-squared:", round(r_value ** 2, 4))
print("p-value:", round(p_value, 6))
print("Standard Error:", round(std_err, 4))

print("\nRegression Equation:")
print(
    f"Respiratory Rate = {intercept:.4f} + "
    f"({slope:.4f} × Pulse)"
)


y_pred = intercept + slope * x

plt.figure(figsize=(8, 5))
plt.scatter(x, y, alpha=0.6)
plt.plot(x, y_pred)

plt.title("Regression Analysis: Pulse vs Respiratory Rate")
plt.xlabel("Pulse")
plt.ylabel("Respiratory Rate")
plt.grid(True, alpha=0.3)

plt.show()
```

```python
# 2.4.1 Feature Selection for K-Means

kmeans_features = [
    'rectal_temp',
    'pulse',
    'respiratory_rate',
    'packed_cell_volume',
    'total_protein',
    'abdomo_protein',
    'nasogastric_reflux_ph'
]


cluster_data = df[kmeans_features].copy()


print("Selected features for K-Means clustering:")
print(kmeans_features)

print("\nShape of clustering dataset:")
print(cluster_data.shape)

print("\nFirst five rows:")
display(cluster_data.head())

print("\nSummary statistics:")
display(cluster_data.describe())
```

```python
# 2.4.2 Standardization


from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()


X_scaled = scaler.fit_transform(cluster_data)


X_scaled_df = pd.DataFrame(
    X_scaled,
    columns=kmeans_features,
    index=cluster_data.index
)

print("Standardized dataset:")
display(X_scaled_df.head())

print("\nMean after standardization:")
display(X_scaled_df.mean().round(4))

print("\nStandard deviation after standardization:")
display(X_scaled_df.std().round(4))
```

```python
# 2.4.3 Elbow Method - WCSS

from sklearn.cluster import KMeans
import matplotlib.pyplot as plt


wcss = []

for k in range(1, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)


wcss_table = pd.DataFrame({
    'Number of Clusters (K)': range(1, 11),
    'WCSS': wcss
})

display(wcss_table)


plt.figure(figsize=(9, 6))

plt.plot(
    range(1, 11),
    wcss,
    marker='o'
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Within-Cluster Sum of Squares (WCSS)")
plt.title("Elbow Method for K-Means Clustering")
plt.xticks(range(1, 11))
plt.grid(True, alpha=0.3)

plt.show()
```

```python
# 2.4.4 Silhouette Method


from sklearn.metrics import silhouette_score

silhouette_scores = []


for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)


silhouette_table = pd.DataFrame({
    'Number of Clusters (K)': range(2, 11),
    'Silhouette Score': silhouette_scores
})

display(silhouette_table)


best_k_silhouette = range(2, 11)[
    silhouette_scores.index(max(silhouette_scores))
]

print("K with highest Silhouette Score:", best_k_silhouette)
print(
    "Highest Silhouette Score:",
    round(max(silhouette_scores), 4)
)


plt.figure(figsize=(9, 6))

plt.plot(
    range(2, 11),
    silhouette_scores,
    marker='o'
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Average Silhouette Score")
plt.title("Silhouette Method for K-Means Clustering")
plt.xticks(range(2, 11))
plt.grid(True, alpha=0.3)

plt.show()
```

```python
# 2.4.5 Final K-Means Clustering - K = 2


from sklearn.cluster import KMeans


k = 2


kmeans_final = KMeans(
    n_clusters=k,
    random_state=42,
    n_init=10
)


cluster_labels = kmeans_final.fit_predict(X_scaled)


clustered_data = cluster_data.copy()
clustered_data['KMeans_Cluster'] = cluster_labels

print("K-Means clustering completed successfully.")

print("\nCluster labels:")
print(clustered_data['KMeans_Cluster'].value_counts().sort_index())

print("\nFirst five observations:")
display(clustered_data.head())
```

```python
# 2.4.6 Euclidean Distance


from scipy.spatial.distance import cdist


distances = cdist(
    X_scaled,
    kmeans_final.cluster_centers_,
    metric='euclidean'
)


distance_df = pd.DataFrame(
    distances,
    columns=['Distance_to_Cluster_0', 'Distance_to_Cluster_1'],
    index=cluster_data.index
)


distance_df['Assigned_Cluster'] = cluster_labels

print("Euclidean distances from observations to cluster centroids:")
display(distance_df.head(10))

print("\nAverage distance to each centroid:")
display(
    distance_df[
        ['Distance_to_Cluster_0', 'Distance_to_Cluster_1']
    ].mean().round(4)
)
```
